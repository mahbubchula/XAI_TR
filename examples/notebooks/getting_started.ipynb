## Part 3: NumPy - Working with Numbers

NumPy is the fundamental package for numerical computing in Python. It's essential for ML!

```python
import numpy as np

# Create arrays (like lists but better for math)
measurements = np.array([120, 80, 98.6, 72])

# Easy math operations on entire arrays
print("Original measurements:", measurements)
print("Doubled:", measurements * 2)
print("Plus 10:", measurements + 10)
print("Mean:", np.mean(measurements))
print("Standard deviation:", np.std(measurements))
```

```python
# Create a 2D array (like a table)
# Each row is a patient, columns are: age, weight, height
patient_data = np.array([
    [25, 70, 175],
    [30, 65, 168],
    [45, 80, 182],
    [50, 75, 170]
])

print("Patient data shape:", patient_data.shape)  # (4 patients, 3 measurements)
print("\nFirst patient:", patient_data[0])        # First row
print("All ages:", patient_data[:, 0])             # First column
print("Average weight:", np.mean(patient_data[:, 1]))
```

## Part 4: Pandas - Working with Data Tables

Pandas is perfect for working with structured data (like spreadsheets).

```python
import pandas as pd

# Create a DataFrame (think of it as an Excel table)
data = {
    'patient_id': [1, 2, 3, 4, 5],
    'age': [25, 30, 45, 50, 35],
    'weight': [70, 65, 80, 75, 68],
    'blood_pressure': [120, 125, 140, 135, 118],
    'has_disease': [0, 0, 1, 1, 0]  # 0=No, 1=Yes
}

df = pd.DataFrame(data)

# Display the data
print("Patient Data:")
print(df)
```

```python
# Useful pandas operations

# Get basic statistics
print("Summary Statistics:")
print(df.describe())

# Access a column
print("\nAges:", df['age'].tolist())

# Filter rows
print("\nPatients over 40:")
print(df[df['age'] > 40])

# Calculate something new
df['bmi'] = df['weight'] / ((1.75) ** 2)  # Simple BMI calculation
print("\nData with BMI:")
print(df)
```

## Part 5: Visualization with Matplotlib

Seeing your data helps understand it better!

```python
import matplotlib.pyplot as plt

# Make plots appear in the notebook
%matplotlib inline

# Create a simple scatter plot
plt.figure(figsize=(10, 6))

# Plot age vs blood pressure, color by disease status
colors = ['green' if d == 0 else 'red' for d in df['has_disease']]
plt.scatter(df['age'], df['blood_pressure'], c=colors, s=100, alpha=0.6)

plt.xlabel('Age', fontsize=12)
plt.ylabel('Blood Pressure', fontsize=12)
plt.title('Age vs Blood Pressure (Green=Healthy, Red=Disease)', fontsize=14)
plt.grid(True, alpha=0.3)

plt.show()
```

## Part 6: Your First Machine Learning Model!

Let's build a simple model to predict disease based on age and blood pressure.

**Goal:** Given a patient's age and blood pressure, predict if they have the disease.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

# Prepare the data
# X = features (what we use to predict)
# y = target (what we want to predict)

X = df[['age', 'blood_pressure', 'weight']].values  # Features
y = df['has_disease'].values                         # Target

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nFeatures (first 3 rows):")
print(X[:3])
print("\nTargets:", y)
```

```python
# Note: In real ML, we'd split data into training and testing sets
# But with only 5 samples, we'll train on all data for demonstration

# Create and train the model
model = DecisionTreeClassifier(random_state=42, max_depth=2)

# Train the model (this is where the learning happens!)
model.fit(X, y)

print("✓ Model trained successfully!")
```

```python
# Make predictions
predictions = model.predict(X)

print("Actual vs Predicted:")
for i in range(len(y)):
    actual = "Disease" if y[i] == 1 else "Healthy"
    predicted = "Disease" if predictions[i] == 1 else "Healthy"
    match = "✓" if y[i] == predictions[i] else "✗"
    print(f"Patient {i+1}: Actual={actual:8s}, Predicted={predicted:8s} {match}")

# Calculate accuracy
accuracy = accuracy_score(y, predictions)
print(f"\nAccuracy: {accuracy:.1%}")
```

```python
# Predict for a new patient!
new_patient = [[40, 145, 78]]  # Age=40, BP=145, Weight=78

prediction = model.predict(new_patient)
probability = model.predict_proba(new_patient)

result = "Disease" if prediction[0] == 1 else "Healthy"
confidence = probability[0][prediction[0]] * 100

print(f"New patient prediction: {result}")
print(f"Confidence: {confidence:.1f}%")
print(f"\nProbabilities: Healthy={probability[0][0]:.2f}, Disease={probability[0][1]:.2f}")
```

## Part 7: Understanding What the Model Learned

This is where Explainable AI (XAI) comes in!

```python
# See which features are most important
feature_names = ['age', 'blood_pressure', 'weight']
importances = model.feature_importances_

print("Feature Importance:")
for name, importance in zip(feature_names, importances):
    print(f"{name:15s}: {importance:.3f} {'█' * int(importance * 50)}")
```

```python
# Visualize feature importance
plt.figure(figsize=(10, 6))
plt.bar(feature_names, importances, color=['#2E86AB', '#A23B72', '#F18F01'])
plt.xlabel('Features', fontsize=12)
plt.ylabel('Importance', fontsize=12)
plt.title('Which Features Matter Most for Prediction?', fontsize=14)
plt.ylim(0, 1)
plt.grid(axis='y', alpha=0.3)
plt.show()
```
